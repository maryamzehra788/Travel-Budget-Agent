```python
# 1. Database - 3 Places
city1 = "Multan"
per_day1 = 7000
car_cost1 = 6000
air_cost1 = 18000

city2 = "Quetta"
per_day2 = 9500
car_cost2 = 14000
air_cost2 = 28000

city3 = "Gwadar"
per_day3 = 13000
car_cost3 = 25000
air_cost3 = 32000

# 2. Input
paisa = int(input("Total budget batao: "))
din = int(input("Trip kitne din ki hai: "))
sawari = input("Kis se jana hai 'car' ya 'plane': ").lower()

# 3. Hisab
karcha1 = per_day1 * din
karcha2 = per_day2 * din
karcha3 = per_day3 * din

if sawari == "car":
    final1 = karcha1 + car_cost1
    final2 = karcha2 + car_cost2
    final3 = karcha3 + car_cost3
elif sawari == "plane":
    final1 = karcha1 + air_cost1
    final2 = karcha2 + air_cost2
    final3 = karcha3 + air_cost3
else:
    print("Error: 'car' ya 'plane' likho")
    quit()

print(f"\n--- {sawari} se mumkin options ---\n")

# 4. Result
option = False

if paisa >= final1:
    print(f"1. {city1} - {din} din: {final1} Rupees")
    option = True

if paisa >= final2:
    print(f"2. {city2} - {din} din: {final2} Rupees")
    option = True

if paisa >= final3:
    print(f"3. {city3} - {din} din: {final3} Rupees")
    option = True

if option == False:
    print("Budget kam hai trip ke liye.")
    lowest = min(final1, final2, final3)
    print(f"Minimum chahiye: {lowest} Rupees")
```
